# 1. Logistic Regression (What, Why and How?)

## a. Abstract Overview
<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">Logistic regression is a statistical model that allows us to predict a categorical label based on historical feature data by predicting the probability of an outcome, event, or observation.</h3>


<img align="center" width="700" src="images/logi6.png"  > 
    


## b. Types of Logistic Regression

<img align="center" width="900" src="images/logi7.png"  > 

## c. Logistic Regression is Linear Model
- In **Regression**, a `Linear Model` means if you plot all the input features and the output variable, then there exist a straight line or hyperplane that rougly estimate the outcome. For single input feature there exist a line, for two input features there exist a plane, and for more than two input features there exist a hyperplane that estimates the outcome. For example $y = 2x+5$ is a linear model, because the output changes by the same amount for any given change in input (slope is constant). On the contrary $y = x^2+3$ is a `Non-Linear Model`, because the output donot change with the same amount for any given change in input (slope is not constant)
- In **Classification**, a `Linear Model` means if you plot all the `n` input features, then there exist a `n-1` dimensional line/hyperplane that can separates different classes. For single input feature there exist a point, for two input features there exist a line, for three input feature there exist a plane, and for more than three input features there exist a hyperplane that separates the datapoints of different classes.

<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">Linearly separable data is data that if graphed in two dimensions, can be separated by a straight line.</h3>
    

    
<img align="center" width="1200" src="images/logi4.png"  > 

## d. Assumptions for Logistic Regression

Four assumptions for **Linear Regression** are `Linearity` (require a linear relationship between the input features and output variable), `Independence` (Input features are independent of eachother), `Homoscedasticity` (variance of the residuals is constants) and `Normality` (residuals are normally distributed). On the contrary **Logistic Regression** does not require `Linearity`, `Homoscedasticity` and `Normality`, however requires `Independence`.
<br><br>
Six assumptions for **Logistic Regression** are:

<font face = "Garamond" color=green size=3>**a. Input features should not be too highly correlated with each other (little or no multicollinearity)** </font>

<font face = "Garamond" color=green size=3>**b. Response variable is binary** </font>

<font face = "Garamond" color=green size=3>**c. Observation are independent of each other, i.e., the observations should not come from repeated measurements or matched data** </font>


<font face = "Garamond" color=green size=3>**d. Prefers large sample size**</font>

<font face = "Garamond" color=green size=3>**e. No extreme outliers** </font>

<font face = "Garamond" color=green size=3>**f. Feature variables are linearly related to log odds** </font>

>- Log odds play an important role in logistic regression as it converts the LR model from probability based to a likelihood based model. More on this later...

##  e. From Linear Regression to Logistic Regression
###   A comparison
**Similarities:**
- Linear Regression and Logistic Regression both are `supervised` Machine Learning algorithms.
- Linear Regression and Logistic Regression, both the models are `parametric` i.e. both the models use linear equations for predictions. In general, every Linear Model consists of sum of parameters (regression coefficients or weights) multiplied by independent variables.
$$y=\beta_0 + \beta_1 x_1 + \beta_2 x_2 + \cdots + \beta_m x_m$$  $$ y=w_0  + w_1 x_1 + w_2 x_2 + \cdots + w_m x_m$$

**Differences:**
- Linear Regression is used to handle `regression problems` whereas Logistic regression is used to handle the `classification problems`.
- Linear regression provides a `continuous output` but Logistic regression provides `discrete output`.
- The purpose of Linear Regression is to find the `best-fitted line` while Logistic regression is one step ahead and fitting the line values to the `sigmoid curve`.
- The method for calculating loss function in linear regression is the` mean squared error` whereas for logistic regression it is `maximum likelihood estimation`.

**Pseudocode:**<br>
```
1. Initialize weights/coefficients with random values
2. Run a loop either a fixed number of times or till convergence
    - Select a random point
    - If (correctly classified)
        - Leave it
    - If (miss-classified)
        - Move the line towards the point
```
<img align="center" width="600" src="images/logi9b.png"  > 

>- If a `negative` point is in the `positive` region you `Subtract`
>- If a `positive` point is in the `negative` region you `Add`


**Algorithm 1:**<br>
1. epochs = 100, $\eta=0.01$<br>
2. for i in epochs:<br>
3. $\hspace{1cm}\text{Pick random } x_i \in P \cup N$<br><br>
4. $\hspace{1cm}\text{if } x_i \in N and \sum_{i=0}^{m} w_ix_i\ge 0$ then $\hspace{1cm}$#A -ve point in +ve region<br>
5. $\hspace{2cm}\vec{w}_{new} = \vec{w}_{old} - \eta \times \vec{x_i}$<br><br>
6. $\hspace{1cm}\text{if } x_i \in P and \sum_{i=0}^{m} w_ix_i\lt 0$ then $\hspace{1cm}$#A +ve point in -ve region<br>
7. $\hspace{2cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times \vec{x_i}$<br>


>- Note: In each iteration, we update the weights only if the randomly selected point is misclassified

<br><br><br>
**Algorithm 2:** A more elegant way of writing above Algo without using the two if conditions<br>
1. epochs = 100, $\eta=0.01$<br>
2. for i in epochs:<br>
3. $\hspace{1cm}\text{Pick random } x_i \in P \cup N$<br>
4. $\hspace{1cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$<br>

**Note:**
- If the data point is correctly classified, then $(y_i - \hat{y}_i)$ will be zero and no update will occur
- If $y_i=1$ and $\hat{y}_i=0$, that means a positive point in negative region, so addition will be done automatically
- If $y_i=0$ and $\hat{y}_i=1$, that means a negative point in positive region, so subtraction will be done automatically


# 3. Logistic Regression using Single Layer Perceptron (`Step Function`)

**Generate Dataset**

````python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(n_samples=100,           
                           n_features=2,            
                           n_classes=2,             
                           n_clusters_per_class=1,  
                           class_sep=20,            
                           random_state=41,
                           n_informative=1,
                           n_redundant=0,
                           hypercube=False)
X.shape, y.shape
````

````python
X[0:5]
````

````python
y
````

**Visualize Dataset**

````python
import matplotlib.pyplot as plt

plt.scatter(x=X[:,0], y=X[:,1], c=y, cmap='winter',  s=100)

plt.title("Classification")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show();
````

**Perceptron Training Algorithm**
$$\large \hspace{0.5cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$$

````python
def perceptron(X,y): 
    X = np.insert(X, 0, 1, axis=1)    # add an additional column of ones in matrix X at column zero
    weights = np.ones(X.shape[1])     # initialize weights vector having size as the no of cols of X containing ones
    lr = 0.1
    epocs = 1000
    for i in range(epocs):
        j = np.random.randint(0,100)                        # select a random index of data point from 0 to 99
        y_hat = step( np.sum( np.dot(X[j], weights) ) )     # calculate y_hat using step function
        weights = weights   +   lr * (y[j] - y_hat) * X[j]  # update the weights vector
    return weights[0],weights[1:]                           # return bias (w0) and the remaining weights (w1,w2)
````

````python
def step(z):
    return 1 if z>=0 else 0
````

````python
# Train the model and get the bias and weights
bias , weights = perceptron(X,y)
print("𝑤0: ", bias)
print("w1 and w2: ", weights)
````

**Visualize Result**
- $Ax+By+C=0$
- Slope $= -\large \frac{A}{B}$
- y-intercept $= -\large \frac{C}{B}$

````python
m = -(weights[0]/weights[1])
b = -(bias/weights[1])

x_input = np.linspace(-3,3,100)
y_input = m*x_input + b

plt.plot(x_input, y_input, color='red', linewidth=3)
plt.scatter(X[:,0],X[:,1], c=y, cmap='winter',  s=100)
plt.ylim(-3,2)
plt.show();
````

# 4. Logistic Regression using Scikit-Learn Model

````python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X, y)
````

````python
print("w0: ", model.intercept_)
print("w1 and w2: ", model.coef_)
````

**Calculating Slope and y-intercept:**
- $Ax+By+C=0$
- Slope $= -\large \frac{A}{B}$
- y-intercept $= -\large \frac{C}{B}$

````python
m = -(model.coef_[0][0]/model.coef_[0][1])
b = -(model.intercept_/model.coef_[0][1])

x_input1 = np.linspace(-3,3,100)
y_input1 = m*x_input + b

plt.plot(x_input,  y_input,  color='red',   linewidth=3) # Resulting Line of Single Layer perceptron
plt.plot(x_input1, y_input1, color='black', linewidth=3) # Resulting Line of Logistic Regression algorithm
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.ylim(-3,2)
plt.show();
````

# 5. Improving Single Layer Perceptron  (`Step Function`) Algorithm

## a. Limitation
**Algorithm:**<br>
epochs = 100, $\eta=0.01$<br>
for i in epochs:<br>
$\hspace{0.5cm}\text{randomly select a point } x_i$<br>
$\hspace{0.5cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$<br>


>- Our perceptron algorithm work only till the time all data points are on the correct side of the line. As soon as it is done the algorithm stops. In simple words, no update occurs if the point picked at random is a correctly classified point, i.e., $y$ and $\hat{y}$ are both either 0 or 1.
>- We need to make changes in the update function:

<font face = "Garamond" color=red size=5>$$\large \hspace{0.5cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$$</font>

## b. How to Overcome this Limitations?

<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">An update should occur for both correctly as well as misclassified points</h3>

- **A `misclassified point pull` the line towards itself:**
   - If a point is `closer` to the line, it will pull the line with `smaller` magnitude
   - If a point is `far` from the line, it will pull the line with `greater` magnitude
- **A `correctly classified point push` the line away from it:**
   - If a point is `closer` to the line, it will push the line with `greater` magnitude
   - If a point is `far` from the line, it will push the line with `smaller` magnitude

## c. How to Implement?

### Old Technique (Using Step Function)
<img align="center" width="800" src="images/logi8c.png"  > 
<br><br>
<font face = "Garamond" color=magenta size=5>$$\large \hspace{0.5cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$$</font>

<font face = "Garamond" color=magenta size=3>**Limitation:** </font> Since the output of step function is discrete, i.e., either 0 or 1, therefore, the update of $\overrightarrow{w}$ occur only if a point is mis-classified </font>

### New Technique (Using Sigmoid Function)
<font face = "Garamond" color=green size=5>$$ \sigma(z) = \large \frac{1}{1+e^{-z}}$$</font>

<p align="center"><div class="alert alert-success" style="margin: 20px"><strong> 
- The Sigmoid function takes any real value as input and outputs another value between 0 and 1 (probability). <br>
- In Sigmoid function, $e$ is the Euler's number, whose value is approximately 2.718, and is the base of the natural logarithms.<br>
- The Sigmoid function is continuously differentiable, and monotonic function having a fixed output range.<br>
</strong></div></p>


<font face = "Garamond" color=magenta size=5>$$\large \hspace{0.5cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$$</font>

<img align="center" width="1000" src="images/logi10a.png"  > 

>- If Z is a very large positive number, $\sigma(z)$ approaches to 1.0
>- If Z is a very large negative number, $\sigma(z)$ approaches to 0.0
>- If Z=0, $\sigma(z)=0.5$
>- By replacing the step function with sigmoid function, the value of $\hat{y}$ will not be discrete, rather will be in the range of 0 and 1.
>- This will solve our problem, as the value of $(y_i - \hat{y_i})$ will not be zero even for the correctly classified points.
>- So if we use sigmoid function, an update will occur in $\vec{w}$ for both correctly and incorrectly classified points. Because $\hat{y}$ is a continous value between 0 and 1. However the magnitude of update will be different.

### Working on Sample Data Points:
- **A misclassified point `pull` the line towards itself**
- **A correctly classified point `push` the line away from it**
<font face = "Garamond" color=magenta size=4>$$\large \hspace{0.5cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$$</font>

## d. Python Code 

**Perceptron Training Algorithm using Sigmoid Function**<br>
$\large \hspace{0.5cm}\vec{w}_{new} = \vec{w}_{old} + \eta \times (y_i - \hat{y}_i)\vec{x_i}$

````python
def perceptron(X, y): 
    X = np.insert(X, 0, 1, axis=1)    # add an additional column of ones in matrix X at column zero
    weights = np.ones(X.shape[1])     # initialize weights vector having size as the no of cols of X containing ones
    lr = 0.1
    epocs = 1000

    for i in range(epocs):
        j = np.random.randint(0,100)                        # select a random index of data point
        y_hat = sigmoid( np.sum( np.dot(X[j], weights) ) )  # calculate y_hat using `sigmoid` function
        weights = weights  +  lr * (y[j] - y_hat) * X[j]    # update the weights vector
    return weights[0],weights[1:]                           # return bias (w0) and the remaining weights (w1,w2)
````

$\large \sigma(z) = \frac{1}{(1+e^{-z})}$

````python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
````

````python
# Train the model and get the bias and weights
bias , weights = perceptron(X, y)
print("w0: ", bias)
print("w1 and w2: ", weights)
````

**Visualize Result**
- $Ax+By+C=0$
- Slope $= -\large \frac{A}{B}$
- y-intercept $= -\large \frac{C}{B}$

````python
m = -(weights[0]/weights[1])
b = -(bias/weights[1])

x_input2 = np.linspace(-3,3,100)
y_input2 = m*x_input + b


plt.plot(x_input,y_input,color='red',linewidth=3)     # Resulting Line of Single Layer perceptron (Step)
plt.plot(x_input1,y_input1,color='black',linewidth=3) # Resulting Line of Logistic Regression algorithm
plt.plot(x_input2,y_input2,color='brown',linewidth=3) # Resulting Line of Single Layer Perceptron (Sigmoid)

plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
plt.show();
````

- Each time you execute the perceptron algorithm using `step` or `sigmoid` function you may get different decision boundry, because you get different weights. The reason is random selection of data point in each iteration. Moreover, the resulting decision boundries are not as good as the one generated by Scikit-Learn's Logistic Regression Model
- The Scikit-Learn's Logistic Regression actually use `Gradient Descent` algorithm to find the optimal weights for a given dataset by minimizing the errors between the predicted probabilities and the true class labels by minimizing `Cross-Entropy-Loss` or `Log-Loss` function.
