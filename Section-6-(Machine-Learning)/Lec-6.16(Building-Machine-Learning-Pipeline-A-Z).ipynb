## a. Missing Value Imputation
- **Why do we need to handle missing values?** 
    - If the missing values are not handled properly, you may end up building a biased machine learning model which will lead to incorrect results.
    - Many machine learning algorithms fail if the dataset contains missing values. However, algorithms like K-nearest and Naive Bayes support data with missing values.

- **How to treat missing values?**
    - Analyze each column with missing values carefully to understand the reasons behind the missing values as it is crucial to find out the strategy for handling the missing values. There are 2 primary ways of handling missing values:
        - Deleting the Missing values: Drop rows (List-wise deletion) having missing values or drop the entire column
        - Imputing the Missing Values: Replace the missing value with a value
            - `Univariate Imputation` 
                - Handling Missing Values using Panda's `fillna()` method
                - Handling Missing Values using sklearn's `SimpleImputer()` transformer
                - Use of Column Transformer
            - `Multivariate Imputation`
                - Handling Missing Values using sklearn's `IterativeImputer()` transformer 
                - Handling Missing Values using sklearn's `KNNImputer()` transformer

**Train-Test Split**

```python
from sklearn.model_selection import train_test_split
X = df.drop('salary', axis=1)
y = df['salary']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
X_train.head()
```

**Apply `SimpleImputer` Transformer on `age` Column**

```python
from sklearn.impute import SimpleImputer
# define imputer
si = SimpleImputer(missing_values=np.nan, strategy='mean')
#train it
si.fit(X_train[['cgpa']]) 

# transform() will return a numpy array which is being assigned to the cgpa column of dataframe
X_train['cgpa'] = si.transform(X_train[['cgpa']])
X_test['cgpa'] = si.transform(X_test[['cgpa']])
X_train.head()
```

```python
#Verify
X_train.isnull().sum()
```

**Apply `OneHotEncoder` Transformer on `gender` Column (Nominal Data)**

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop='first', sparse=False, dtype=np.int32)
#train
ohe.fit(X_train[['gender']])

# transform() will return a numpy array which is being assigned to the gender column of dataframe
X_train['gender'] = ohe.transform(X_train[['gender']])
X_test['gender'] = ohe.transform(X_test[['gender']])
#Verify
X_train.head()
```

**Apply `OrdinalEncoder` Transformer on `iq` Column (Ordinal Data)**

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[['Low', 'Medium', 'High']], dtype=np.int8)
#train
oe.fit(X_train[['iq']])
# transform() will return a numpy array which is being assigned to the gender column of dataframe
X_train['iq'] = oe.transform(X_train[['iq']])
X_test['iq'] = oe.transform(X_test[['iq']])
#verify
X_train.head()
```

**Instentiate and Train `LinearRegression()` Model**

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train,y_train)
```

**Predict User Input**

```python
test_input = np.array(['Male', 3.5, 'High']).reshape(1,3)
# Transform above user input to be fed to ML model 
test_input_transformed = np.array([1, 3.5, 2]).reshape(1,3)
test_input, test_input_transformed
```

```python
import warnings
warnings.filterwarnings('ignore')
model.predict(test_input_transformed)
```

## Example 2: Repeat Example 1 using Scikit-Learn's `ColumnTransformer` Class
Visit ColumnTransformer Class in Scikit-Learn:https://scikit-learn.org/stable/modules/generated/sklearn.compose.ColumnTransformer.html

<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">ColumnTransformer allows to apply different Transformers on different columns of our dataset in a simple and elegant way</h3>

**Load Dataset**

```python
import pandas as pd
import numpy as np
df = pd.read_csv('datasets/salary-package1.csv')
df
```

**Train-Test Split**

```python
from sklearn.model_selection import train_test_split
X = df.drop('salary', axis=1)
y = df['salary']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
```

```python
X_train
```

**Create `ColumnTransformer` Object, Train and Transform**

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer

col_tfm = ColumnTransformer(
                        transformers=[
                                    ('tfm1', OneHotEncoder(drop='first',sparse=False),    ['gender']),
                                    ('tfm2', SimpleImputer(missing_values=np.nan, strategy='mean'),  ['cgpa']),
                                    ('tfm3', OrdinalEncoder(categories=[['Low', 'Medium', 'High']]), ['iq'])
                                     ],
                        remainder='passthrough')
type(col_tfm)
```

```python
# Train the ColumnTransformer object
col_tfm.fit(X_train)
```

```python
# Call transform on the trained ColumnTransformer object
# Return the transformed numPy array having three columns gender, cgpa, iq
arr_Xtrain = col_tfm.transform(X_train)
arr_Xtest = col_tfm.transform(X_test)
arr_Xtest
```

>- ColumnTransformer returns a numPy array even if we input a DataFrame object which makes it difficult to track the columns, i.e., we cannot track columns by name rather we have to track them by index.

**Instentiate and Train `LinearRegression()` Model**

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(arr_Xtrain, y_train)
```

**Predict**

```python
test_input = np.array([['Male', 3.5, 'High']])
test_input_transformed = np.array([[1, 3.5, 2]])
test_input
```

```python
test_input = np.array(['Male', 3.5, 'High']).reshape(1,3)
test_input_transformed = np.array([1, 3.5, 2]).reshape(1,3)
test_input
```

```python
import warnings
warnings.filterwarnings('ignore')
model.predict(test_input_transformed)
```

**Limitations of `ColumnTransformer`:**
- In a ColumnTransformer we cannot apply multiple transforms to a single column.
- ColumnTransformer returns a numPy array even if we input a DataFrame object which makes it difficult to track the columns, i.e., we cannot track columns by name rather we have to track them by index.


# 3. Overview of Machine Learning Pipeline

## What is a Machine Learning Pipeline?
<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">Pipeline is a mechanism that chains together multiple steps together so that the o/p of each step is used as i/p to the next step</h3>
    
<img align="center" width="1000" src="images/pipeline1.png"  > 
    
<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">A machine learning pipeline is simply a set of steps that you follow while working on your project. <br>This could include things like acquiring and preprocessing your data, training, hypertuning and validating models, and finally deploying them to make predictions. </h3>
 

>- The sequence of transformers does matter, as the output of each transformer become the input of next transformer
>- Moreover, there will be an estimator object at the end of the pipeline
>- Transformers can be built-in Scikit-Learn Transformer classes, or you can built your own customized transformers by inheriting that class from appropriate base transformer and base estimator classes.

**Train-Test Split**

```python
from sklearn.model_selection import train_test_split
X = df.drop('sales', axis=1)
y = df['sales']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=54)
```

**Create Pipeline Object, Train and Predict** <img align="center" width="600" src="images/pipeline2.png"  >  

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

# steps argument of Pipeline() is a list of two valued tuples 
# unlike transformers argument of ColumnTransformer (3 valued tuple)
# SimpleImputer and StandardScaler will be applied to all the columns of the dataset
pipe = Pipeline(
               steps=[
                     ('si',  SimpleImputer(missing_values=np.nan, strategy='mean')),
                     ('ss',  StandardScaler()),
                     ('lr',  LinearRegression())
                    ]
                )
pipe
```

```python
# To graphically view what the pipeline object will do
from sklearn import set_config
set_config(display='diagram')
pipe
```

**Train the Pipeline instead of Model**
- On a pipeline object, when you call the `fit()` method (by passing it raw or unprocessed training data), it will internally call the `fit_transform()` method of all transformers one after the other and transform the data. The transformed data is then passed to the final estimator which calls it's `fit()` method.

```python
pipe.fit(X_train, y_train)
```

**Predict using the Pipeline instead of Model**
- On a trained pipeline object, when you call the `predict()` method (by passing raw or unprocessed test data), it will internally call only the `transform()` method of all transformers one after the other and transform the data.  The transformed test data is then passed to the final estimator which calls it's `predict()` method.

```python
pipe.predict(X_test)
```

**Evaluate using the Pipelined Model**

```python
pipe.score(X_test, y_test)
```

```python
# You can call methods on a specific transformer of this trained pipeline object
# Let us call fit_transform() on SimpleImputer, will return a NumPy array after imputing the missing values
pipe['si'].fit_transform(X_train)
```

```python
# You can call methods on a specific transformer of this trained pipeline object
# Let us call fit_transform() on StandardScalar, will return a NumPy array after scaling the data
pipe['ss'].fit_transform(X_train)
```

## Example 4: Using ColumnTransformer in a Pipeline

- A transformer in a pipeline is applied on the entire input dataset or the o/p of the previous transformer. 
- We can have missing values in numeric variables as well as missing values in categorical variables, so we have to use different imputation techniques on numeric columns and different techniques on categorical columns.
- Moreover we have to apply different encoding techniques to categorical columns, for nominal data we apply OneHotEncoder while for ordinal data we apply OrdinalEncoder
- At times we want to apply different transformation to different columns of our dataset. For this we can create multiple pipelines and later combine them together
    - For processing numeric data we can create a separate pipeline
    - For processing categorical data we can create a separate pipeline
    - Later we can combine these two pipelines to a single one, and later any transformer that we place in the pipeline will be applied to the entire output dataset of the previous step of pipeline :)

### Create Pipeline Object, Train and Predict

**Create Two Pipeline Objects to handle Categorical and Numeric Data**

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer 
from sklearn.preprocessing import OneHotEncoder

# Preprocess categorical columns ("sex", "smoker", "day", "time")
categorical_transformer = Pipeline(steps=[
                                     ('imputer', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),
                                     ('onehot', OneHotEncoder(drop='first', handle_unknown='ignore'))
                                   ])

# Preprocess numeric columns ("total_bill", "size")
numeric_transformer = Pipeline(steps=[
                                      ('imputer', SimpleImputer(missing_values=np.nan, strategy='mean'))  
                                  ])
```

**Combine above two Pipelines using ColumnTransformer**

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
                              transformers=[
                                           ('categorical', categorical_transformer, [1,2,3,4]),
                                           ('numerical', numeric_transformer, [0,5])
                                           ]
                                )
```

**Create the Final Pipeline Object**

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

model = Pipeline(steps=[
                        ('preprocessor', preprocessor),
                        ('scaling', StandardScaler()),
                        ('regressor', LinearRegression())
                       ])

model
```

```python
X_train
```

**Train and Evaluate the Pipeline Object**

```python
# Note: when fit() is called on a Pipeline object, internally 
# it will call the fit_transform() method on all the transformers and fit() on the estimator

model.fit(X_train, y_train)
```

```python
# Note: when score() or predict() is called on a Pipeline object, internally 
# only the transform() method is called on all the transformers and score() / predict() on the estimator
model.score(X_test, y_test)
```

**Predict on new input**

```python
total_bill = 100
sex = 'Male'
smoker = 'Yes'
day = 'Sun'
time = 'Dinner'
size = 4
test_input = np.array([[total_bill, sex, smoker, day, time, size]], dtype='object')
test_input
```

### Putting all the Code in one Cell

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
import warnings
warnings.filterwarnings('ignore')

# Read the data
df = pd.read_csv('datasets/tips-missingdata.csv')
# Drop the rows with no labels
df.dropna(axis=0, how='any', subset=['tip'], inplace=True)
# Split data into X (data) & y (labels) and do a train-test split
X = df.drop('tip', axis=1)
y = df['tip']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=54)
X_train


# Preprocess categorical columns ("sex", "smoker", "day", "time")
categorical_transformer = Pipeline(steps=[
                                     ('imputer', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),
                                     ('onehot', OneHotEncoder(drop='first', handle_unknown='ignore'))
                                   ])
# Preprocess numeric columns ("total_bill", "size")
numeric_transformer = Pipeline(steps=[
                                      ('imputer', SimpleImputer(missing_values=np.nan, strategy='mean'))  
                                  ])
# Create a column transformer which combines all of the other transformers into one step
preprocessor = ColumnTransformer(
                              transformers=[
                                           ('categorical', categorical_transformer, [1,2,3,4]),
                                           ('numerical', numeric_transformer, [0,5])
                                           ]
                               )         
# Create the final pipeline
model = Pipeline(steps=[('preprocessor', preprocessor),
                        ('scaling', StandardScaler()),
                        ('regressor', LinearRegression())])

# Fit the model on the training data 
model.fit(X_train, y_train)

print("R2 Score:", cross_val_score(model,X_train, y_train, cv=5, scoring='r2').mean())
test_input = np.array([[100, 'Male', 'Yes', 'Sun', 'Dinner', 4]], dtype='object')
print("Predicted tip amount: ", model.predict(test_input))
```
