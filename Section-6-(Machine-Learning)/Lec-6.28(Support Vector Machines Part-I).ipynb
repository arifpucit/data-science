# 1. Support Vector Machine (A Big Picture)
## a. An Overview
<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">Support Vector Machine (SVM) is a supervised machine learning algorithm, which can be used for both classification and regression problems.</h3>
    
    
<img align="center" width="600" src="images/svm0.png"  >

| Logistic Regression | Support Vector Machine |
| :----------- | :----------- |
| Used for classification alone | Used for classification as well as regression |
| `Probabilistic classifier` that o/p probability estimates for `k` classes and assigns the label of the class with the highest probability| `Deterministic classifier` that o/p a hard label without providing the probability estimate for `k` classes|
| Use statistical properties of data | Use geometric properties of data and is also called maximum margin classifier |
| Identifies a single hyperplane by using the idea of perceptron logic | Identifies three hyperplanes (extends the idea of perceptron logic)|
| Logistic Regression is a linear model and work on linear data only | SVM is a linear model but can work on non-linear data as well (kernel trick)|
| Sensitive to outliers and prone to overfitting | Robust to outliers and risk of overfitting is relatively lower|

## b. Linearly Separable Data

<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">SVM is a linear model, however, work on non-linear datasets as well by using techniques like kernel tricks.</h3>
    

    
<img align="center" width="1200" src="images/logi4.png"  > 

<br><br>    

- **Linear SVM:** When the data is perfectly linearly separable only then we can use Linear SVM. Perfectly linearly separable means that the data points can be classified into 2 classes by using a single straight line (if 2D).
- **Non-Linear SVM:** When the data is not linearly separable then we can use Non-Linear SVM, which means when the data points cannot be separated into 2 classes by using a straight line (if 2D) then we use some advanced techniques like `kernel tricks` to classify them. In most real-world applications we do not find linearly separable datapoints hence we use kernel trick to solve them.

## e. How does SVM Work (A Geometric Intuition )?

<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">SVC is also called Maximum Margin Classifier as it tries to maximize the margin that separates the two classes</h3>    

<img align="center" width="1000" src="images/svm2f.png"  > 

**Support Vectors:**
- Support vectors are the data points that are `closest` to the `hyperplane` and have the greatest impact on the position of the hyperplane. These points are the most critical points in determining the position of the hyperplane and thats why the algorithm is named as Support Vector Machine.
- It's the support vectors that are used to define the hyperplane. 
- SVM algorithm finds the hyperplane that `maximizes` the distance between the support vectors of the different classes. 

**Margin:**
- The margin in SVM is the distance between the hyperplane and the closest data points from each class. It is a measure of how well the hyperplane separates the different classes in the data. 
- The goal of the SVM algorithm is to find the hyperplane that maximizes the margin, as a large margin generally indicates a better separation of the classes and a more robust model.
- The margin is a key concept in SVM, as it helps to increase the model's generalization ability by maximizing the distance between the classes. This helps to reduce the chances of overfitting and also increase the robustness of the model.
    
    
<font face = "Garamond" color=green size=4>**Characteristics of Best Decision Boundary:**
1. <font face = "Garamond" color=green size=3>**Maximizes the margin.**</font>

2. <font face = "Garamond" color=green size=3>**Correctly classifies majority of the data points.**</font>

3. <font face = "Garamond" color=green size=3>**Robust to outliers and noise.**</font>

4. <font face = "Garamond" color=green size=3>**High generalization ability.**</font>

<font face = "Courier New" color=blue size=4>$\large \hspace{2cm}  \overrightarrow{ w}  \overrightarrow{ x}  + w_0 = 0$</font>

**Interpretation of Equation of Vector $\overrightarrow{\rm w}$**
- Since the dot product of $\overrightarrow{ w}$ and $\overrightarrow{x}$ is zero, that means the angle between $\overrightarrow{ w}$ and $\overrightarrow{x}$ is 90 degrees, i.e., they are orthogonal vectors
- Therefore, **$\overrightarrow{ w}$ is a vector which will always be perpendicular to the hyperplane (decision boundry).**
<br><br>
<font face = "Courier New" color=blue size=4>$\large \hspace{2cm}  w^T x  = w \cdot x = |w| \hspace{0.2 cm} |x| \hspace{0.2 cm} cos \theta = 0$</font>

## c. Positive and Negative Hyperplanes
<img align="center" width="700" src="images/svm5b.png"  > 

>- Note the positive and the negative hyperplanes are parallel to the decision boundry, i.e., having same slope but different y-intercepts
>- By increasing $C$  in $Ax+By+C=0$ or $w_0$ in $\overrightarrow{ w}  \overrightarrow{ x}  + w_0 = 0$, the y-intercept decreases. (and vice-versa)

<img align="center" width="500" src="images/svm8b.png"  > 

>- By multiplying $\hspace{0.2cm}w^T   x  + b = +1,\hspace{0.2cm}$ with a factor greater than one (e.g., 2) the decision boundry moves up
>- By multiplying  $\hspace{0.2cm}w^T   x  + b = -1,\hspace{0.2cm}$ with a factor less than one (e.g., 0.5) the decision boundry moves down

<br>

To find the equation of a positive hyperplane, we multiply $\hspace{0.2cm}w^T   x  + b = +1,\hspace{0.2cm}$ with a factor greater than 1 till it hit some positive support vector(s). Similarly to find the equation of a negative hyperplane, we multiply $\hspace{0.2cm}w^T   x  + b = -1,\hspace{0.2cm}$ with a factor less than 1 till it hit some negative support vector(s).</font>

## d. Decision Rule for SVM
- General Rule to `Classify/Predict` a New Data Point:
<font face = "Courier New" color=black size=4>
$$
\large \hat y_i = 
\Bigg\{
\begin{matrix}
+1, & \text{if}& \overrightarrow{w}\overrightarrow{x_i}+b\ge 0 \\
-1, & \text{if} & \overrightarrow{w}\overrightarrow{x_i}+b\lt 0  \\
\end{matrix}
$$
</font>
<br><br>
- **Constraint for `Hard SVM`:** 
    - We need to maximize the margin, such that 
        - NO training point is `below` the `Positive Hyperplane`: <font face = "Courier New" color=green size=3>$\large \hspace{2cm}  \overrightarrow{ w}  \overrightarrow{ x}  + b \ge 1$</font>
        - NO training point is be `above` the `Negative Hyperplane`: <font face = "Courier New" color=blue size=3>$\large \hspace{2cm}  \overrightarrow{ w}  \overrightarrow{ x}  + b \lt -1$</font>
    - Simplify the above two equation to one, assuming that the label of all the positiv training points is $+1$ and the label of all the negative training points is $-1$:<br><br>
<font face = "Courier New" color=magenta size=5>$$\large \hat y_i = y_i(\overrightarrow{w}  \overrightarrow{x_i}  + b) \ge 1$$</font>

## e. Formula to Calculate the Margin


<img align="right" width="400" src="images/svm9b.png"  > 

- The shortest distance between $x_1$ and $x_2$ will be perpendicular to both the positive and negative hyperplanes.
- Now we do have a $\vec{w}$, which is perpendicular to both rather all the three hyperplanes, whose unit vector (vector having a magnitude of one) can be obtained by dividing $\vec{w}$ by its magnitude, i.e., $\lVert w \rVert$. 
- Moreover the margin (d) is actually the projection of ($\vec{\vec{x_2} - \vec{x_1}}$) on $\vec{w}$, So

<font face = "Courier New" color=black size=5>
$\hspace{2 cm} d = (x_2 - x_1).\frac{\vec{w}}{\lVert w \rVert}$

$\hspace{2 cm} d = \frac{x_2\vec{w} - x_1\vec{w}}{\lVert w \rVert}$

$\hspace{2 cm}d = \frac{\vec{w} x_2 - \vec{w} x_1}{\lVert w \rVert} \hspace{2 cm}\cdots \hspace{1 cm}(i)$
</font>
- We already know that for all support vectors <font face = "Courier New" color=magenta size=4>$y_i(\vec{w}\vec{x_i}+b)=1$ </font> holds. 
    - Since $x_2$ is a positive support vector having $y_i$ of $+1$, so <font face = "Courier New" color=green size=4>$\vec{w}\vec{x_2}=1-b$ </font>
    - Since $x_1$ is a negative support vector having $y_i$ of $-1$, so <font face = "Courier New" color=blue size=4>$\vec{w}\vec{x_1}=-1-b$ </font>
<br><br>
- Substituting these values in equation (i):<br><br>

<font face = "Courier New" color=black size=5>
$\hspace{6.5 cm} d = \large \frac{1 - b -(- 1 - b)}{\lVert w \rVert}$</font><br><br>
<font face = "Courier New" color=black size=5>
$\hspace{6.5 cm} d = \large \frac{2}{\lVert w \rVert}  \hspace{2 cm}\cdots \hspace{1 cm}\small(ii)$
</font>

> Margin between the positive and negative hyperplane is two divided by magnitude of $\vec w$, i.e. $\lVert w \rVert$

## f. Mathematical Formulation of SVM
<font face = "Courier New" color=magenta size=6>
$$
argmax_{(w^*, b^*)} \hspace{0.3cm} \frac{2}{\lVert w \rVert}, \text{such that} \hspace{1 cm} y_i(\overrightarrow{w}  \overrightarrow{x_i}  + b) \ge 1
$$
</font>
<br><br>

- Since 2 is a constant, so we can ignore it and the above expression can be writen as:
<font face = "Courier New" color=magenta size=6>
$$
argmax_{(w^*, b^*)} \hspace{0.3cm} \frac{1}{\lVert w \rVert}, \text{such that} \hspace{1 cm} y_i(\overrightarrow{w}  \overrightarrow{x_i}  + b) \ge 1
$$
</font>
<br><br>

- To change above into a minimization problem, we can write as:
<font face = "Courier New" color=blue size=6>    
$$
argmin_{(w^*, b^*)} \hspace{0.3cm} \lVert w \rVert, \text{such that} \hspace{1 cm} y_i(\overrightarrow{w}  \overrightarrow{x_i}  + b) \ge 1
$$
</font> 
<br><br>

- For further ease of calculation, we can write above as:
<font face = "Courier New" color=blue size=6>    
$$
argmin_{(w^*, b^*)} \hspace{0.3cm} \frac{1}{2} \lVert w \rVert ^2, \text{such that} \hspace{1 cm} y_i(\overrightarrow{w}  \overrightarrow{x_i}  + b) \ge 1
$$
</font>
    

### 100\$ Question:

<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">How to find the values of $\overrightarrow{w}$ and $b$ for the best decision boundry that maximizes the distance between the positive and negative hyperplanes?</h3>

<h3 align="center"><div class="alert alert-success" color=magenta style="margin: 20px">One of the way to solve the above constraint optimisation problem is to add one more variable (called Lagrange multiplier) for each constraint and solve it.</h3>

## Langrange Multipliers:
- In mathematical optimization, the method of Lagrange multipliers is a strategy for finding the local maxima and minima of a function subject to equation constraints (i.e., subject to the condition that one or more equations have to be satisfied exactly by the chosen values of the variables). It is named after the mathematician Joseph-Louis Lagrange.
- In the general case, for two functions $f(x)$ and $g(x)$ and Lagrange multiplier ($\large \alpha$ or $\large \lambda$), the Langrangian function is defined as:

<font face = "Courier New" color=green size=6>
$$\hspace{1 cm}L(x,\alpha) = f(x) + \alpha . g(x)$$
</font>

- Where, $\large \lambda$ or $\large \alpha$ is called the Langrange multiplier
<br><br>

- So the above function with constraint can be written as a Langrangian function as shown below:
<font face = "Courier New" color=green size=6>
$$\hspace{1 cm}L = \frac{1}{2}  \lVert w \rVert^2 - \sum \alpha_i [y_i (\overrightarrow{w} \overrightarrow{x_i}+b)-1]$$
</font>

- In order to find the minimum of a constrained function all partial derivatives should be zero.
<img align="center" width="600" src="images/langrange1.png"  > 

- Plug in the value of $\overrightarrow{w}$ in above Loss function, shown in green:
<br><br>
<font face = "Courier New" color=black size=5>
$$\hspace{1 cm}L = \frac{1}{2} \large(\small \sum \alpha_i y_i x_i\large).(\small \sum \alpha_i y_i x_i\large)\hspace{0.3cm}-\hspace{0.3cm} (\small \sum \alpha_i y_i x_i\large).(\small \sum \alpha_j y_j x_j\large) \hspace{0.3cm}-\hspace{0.3cm} b\small \sum \alpha_iy_i + \small \sum \alpha_i$$
</font>

<font face = "Courier New" color=green size=6>
$$\hspace{1 cm}L = \sum \alpha_i - \frac{1}{2} \small \sum \sum \alpha_i \alpha_j y_i y_j x_i.x_j$$
</font>

>- <font face = "Garamond" color=green size=3>**The nice thing about the above objective function is that we have an expression for $\vec w$ in terms of Lagrange multipliers involving no $\vec w$ term.**</font>

**Load the Dataset:**

```python
import numpy as np
import pandas as pd

X = np.array([[1, 4], [2, 5], [3, 5], [3, 4], [6,1], [4,0], [5,2], [5,1]])
y = np.array([1, 1, 1, 1, -1, -1, -1, -1])
df = pd.concat([pd.DataFrame(data = X, columns = ['x1', 'x2']),    pd.Series(y, name ='y')],  axis=1)
df
```

**Visualize Dataset using Scatter Chart:**

```python
import matplotlib.pyplot as plt
plt.scatter(X[:, 0], X[:, 1], c=y, s=100, cmap='winter');
```

**Create an Instance of SVC Model:**

```python
from sklearn.svm import SVC
model = SVC(kernel='linear', C=1)
```

**Train the Model:**

```python
model.fit(X, y)
```

**Check values of different attributes:**

```python
# Get the support vectors using the support_vectors_ attribute of model
model.support_vectors_
```

```python
# get indices of support vectors
model.support_
```

```python
# get count of negative and positive support vectors for each class
model.n_support_
```

```python
# use the coef_ attribute to access the normal vector (w) of the hyperplane
# use the intercept_ attribute to access the bias (b)
w = model.coef_[0]
b = model.intercept_[0]

print('Vector (w) = ', w)
print('Bias (b) = ', b)
```

**Create the Equation of Decision Boundry:**
- $Ax + By + C = 0$<br>
- $y = mx + c$

```python
# For the general form of the equation Ax+By+C = 0, calculate the slope and y-intercept
A = w[0]
B = w[1]
C = b

m = -A/B # slope m = -A/B
c = -C/B # y-intercept is c = -C/B
print("Slope: ", m)

print("y-intercept: ", c)
```

**Plot the Hyperplane/Decision Boundry:**

```python
xx = np.linspace(0, 7, 50)         # Generate 50 evenly spaced numbers between 0 and 6
yy = m * xx + c                    # Calculate corresponding 50 points for output y using the line equation y=mx+c
plt.scatter(X[:, 0], X[:, 1], c=y, s=100, cmap='winter'); # use scatter scatter() to plot the data points
plt.plot(xx, yy);                  # use plot() method to draw the line (Decision boundry)
```

**Plot the Positive, Negative Hyperplanes and Decision Boundry:**

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# User defined plotting function
def plot_svc_decision_function(model, ax=None, plot_support=True):
    """Plot the decision function for a 2D SVC"""
    if ax is None:
        ax = plt.gca()
    xlim = ax.get_xlim()
    ylim = ax.get_ylim()
    
    # create grid to evaluate model
    x = np.linspace(xlim[0], xlim[1], 30)
    y = np.linspace(ylim[0], ylim[1], 30)
    Y, X = np.meshgrid(y, x)
    xy = np.vstack([X.ravel(), Y.ravel()]).T
    P = model.decision_function(xy).reshape(X.shape)
    
    # plot decision boundary and margins
    ax.contour(X, Y, P, colors='k',
               levels=[-1, 0, 1], alpha=0.5,
               linestyles=['--', '-', '--'])
    
    # plot support vectors
    if plot_support:
        ax.scatter(model.support_vectors_[:, 0],
                   model.support_vectors_[:, 1],
                   s=300, linewidth=1, facecolors='none');
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, s=100, cmap='winter');
plot_svc_decision_function(model);
```

**Carry out Prediction on New Data Points:**

```python
point1 = [2,3]
point2 = [4,1]

print(f'Point {point1} belong to class: {model.predict([point1])[0]}')
print(f'Point {point2} belong to class: {model.predict([point2])[0]}')
```

>- Remember, during prediction the model use the decision boundry and the positive and negative hyperplanes has no role to play.
