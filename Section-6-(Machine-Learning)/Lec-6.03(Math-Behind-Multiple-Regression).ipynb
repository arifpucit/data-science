## a. Overview of Multiple Regression
<img align="center" width="900" src="images/simple-multiple-regression1.png"  >
<font face = "Courier New" color=red size=4>$\hspace{5cm} \large \hat{y} = \beta_0 + \beta_1 x_1$</font><br>
<font face = "Courier New" color=red size=4>$\hspace{5cm}\large \hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2$</font><br>
<font face = "Courier New" color=red size=4>$ \hspace{5cm}\large \hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_3$</font><br>
<font face = "Courier New" color=red size=4>$ \hspace{5cm}\large \vdots$</font><br>

<font face = "Courier New" color=green size=5>$ \hspace{5cm}\large \hat{y} = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + \beta_3 x_3 +\cdots + \beta_m x_m$</font><br>
<font face = "Courier New" color=magenta size=5>$$ \hspace{5cm}\large \hat{y} = \beta_0 + \sum_{i=1}^m \beta_i x_i$$</font><br>

- Where, 
    - $\hat{y}$ is the predicted dependent or outcome or response variable.
    - $x_1, x_2, x_3, \cdots, x_m$ are the independent or feature or predictor variables.
    - $\beta_0$ is the minimum value of $y$ when all the feature variables are zero (constant term).
    - $\beta_1, \beta_2, \beta_3, \cdots, \beta_m$  are the values that quantify the effect of respective independent variables on the output or dependent variable.

# 4. How to Do Multiple Linear Regression using Matrices?

## a. Formulation of Multiple Linear Regression using Matrix Representation

<font face = "Courier New" color=magenta size=5>$$ \hspace{5cm}\large \hat{y} = \beta_0 + \sum_{i=1}^m \beta_i x_i \hspace{6cm}$$</font><br>


$$ \large \hat{y_1} = \beta_0 + \beta_1 X_{1,1} + \beta_2 X_{1,2} + \beta_3 X_{1,3} +\cdots + \beta_m X_{1,m}$$
$$ \large \hat{y_2} = \beta_0 + \beta_1 X_{2,1} + \beta_2 X_{2,2} + \beta_3 X_{2,3} +\cdots + \beta_m X_{2,m}$$
$$ \large \hat{y_3} = \beta_0 + \beta_1 X_{3,1} + \beta_2 X_{3,2} + \beta_3 X_{3,3} +\cdots + \beta_m X_{3,m}$$

$$\large \vdots$$

$$ \large \hat{y_n} = \beta_0 + \beta_1 X_{n,1} + \beta_2 X_{n,2} + \beta_3 X_{n,3} +\cdots + \beta_m X_{n,m}$$

- The above overdetermined system of linear equations with n equations and m+1 unknowns can be written in matrix form as:

$$\large 
\begin{bmatrix} \hat{y_1} \\ \hat{y_2} \\ \hat{y_3} \\ \vdots \\ \hat{y_n} \end{bmatrix}\hspace{0.5cm} =\hspace{0.5cm} \begin{bmatrix}
\beta_0 +\beta_1X_{1,1} + \beta_2X_{1,2} + \beta_3X_{1,3} +\cdots + \beta_mX_{1,m} \\ 
\beta_0 + \beta_1X_{2,1} + \beta_2X_{2,2} + \beta_3X_{2,3} +\cdots + \beta_mX_{2,m} \\ 
\beta_0 + \beta_1X_{3,1} + \beta_2X_{3,2} + \beta_3X_{3,3} +\cdots + \beta_mX_{3,m} \\ 
\vdots    \\ 
\beta_0 + \beta_1X_{n,1} + \beta_2X_{n,2} + \beta_3X_{n,3} +\cdots + \beta_mX_{n,m} \\ \end{bmatrix}$$

$$\large 
\begin{bmatrix} \hat{y_1} \\ \hat{y_2} \\ \hat{y_3} \\ \vdots \\ \hat{y_n} \end{bmatrix}\hspace{0.5cm} =\hspace{0.5cm}
\begin{bmatrix}
1 & X_{1,1} & X_{1,2} & X_{1,3} &\cdots & X_{1,m} \\ 
1 & X_{2,1} & X_{2,2} & X_{2,3} &\cdots & X_{2,m} \\ 
1 & X_{3,1} & X_{3,2} & X_{3,3} &\cdots & X_{3,m} \\ 
\vdots &\vdots &\vdots &\vdots &\ddots &\vdots\\ 
1 & X_{n,1} & X_{n,2} & X_{n,3} &\cdots & X_{n,m} \\  \end{bmatrix}
\begin{bmatrix} \beta_0 \\ \beta_1 \\ \beta_2 \\ \vdots \\ \beta_m \end{bmatrix}\hspace{2cm}
$$

<font face = "Courier New" color=red size=6>$$ \hspace{.6cm}\large \hat{Y}_{\small n\times 1} \hspace{.5cm}=\hspace{.5cm} X_{\small n\times \small (m+1)}\hspace{1cm} \beta_{\small (m+1)\times 1}$$</font>

**For a single input feature:**
$$\large 
\begin{bmatrix} \hat{y_1} \\ \hat{y_2} \\ \hat{y_3} \\ \vdots \\ \hat{y_n} \end{bmatrix}\hspace{0.5cm} =\hspace{0.5cm}
\begin{bmatrix}
1 & X_{1,1}  \\ 
1 & X_{2,1}  \\ 
1 & X_{3,1}  \\ 
\vdots &\vdots \\ 
1 & X_{n,1}   \end{bmatrix}
\begin{bmatrix} \beta_0 \\ \beta_1 \end{bmatrix}
$$

**For two input features:**
$$\large 
\hspace{2cm}\begin{bmatrix} \hat{y_1} \\ \hat{y_2} \\ \hat{y_3} \\ \vdots \\ \hat{y_n} \end{bmatrix}\hspace{0.5cm} =\hspace{0.5cm}
\begin{bmatrix}
1 & X_{1,1} & X_{1,2}  \\ 
1 & X_{2,1} & X_{2,2} \\ 
1 & X_{3,1} & X_{3,2} \\ 
\vdots &\vdots &\vdots\\ 
1 & X_{n,1}  & X_{n,2} \end{bmatrix}
\begin{bmatrix} \beta_0 \\ \beta_1 \\ \beta_2 \end{bmatrix}
$$

## b. Intuitive Understanding of the Cost Function for Multiple Linear Regression

<img align="right" width="400" src="images/mlr-errors.png"  >

<br><br><br>
$$\large y \hspace{.5cm}=\hspace{.5cm} \beta_0 + \beta_1 X_1 + \beta_2 X_2 + E(\beta_0, \beta_1, \beta_2 )$$<br>

$$\large y \hspace{.5cm}=\hspace{.5cm} \hat{y} \hspace{1cm}+\hspace{1cm} E(\beta_0, \beta_1, \beta_2 )\hspace{2cm}$$<br>


<font face = "Courier New" color=magenta size=5>$$\large E(\beta_0, \beta_1, \beta_2) \hspace{.5cm}=\hspace{.5cm} y - \hat{y}\hspace{12cm}$$</font><br>


## c. Derivation of  $\beta_0, \beta_1, \beta_2, \cdots,\beta_m$ for MLR using OLS Method
- System of Linear equations for the residual errors can be written as:

$$ \large e_1 = y_1 - \hat{y_1}$$
$$ \large e_2 = y_2 - \hat{y_2}$$
$$ \large e_3 = y_3 - \hat{y_3}$$
$$\large \vdots$$
$$ \large e_n = y_n - \hat{y_n}$$

- The above equations can be written in matrix form:

$$\large  E \hspace{.5cm}=\hspace{.5cm} 
\begin{bmatrix} e_1 \\ e_2 \\ e_3 \\ \vdots \\ e_n \end{bmatrix} \hspace{.5cm}=\hspace{.5cm}
\begin{bmatrix} y_1 \\ y_2 \\ y_3 \\ \vdots \\ y_n \end{bmatrix} - \begin{bmatrix} \hat{y_1} \\ \hat{y_2} \\ \hat{y_3} \\ \vdots \\ \hat{y_n} \end{bmatrix} \hspace{.5cm}=\hspace{.5cm}
\begin{bmatrix} y_1 - \hat{y_1} \\ y_2 - \hat{y_2} \\ y_3 - \hat{y_3} \\ \vdots \\ y_n - \hat{y_n} \end{bmatrix}
$$

<font face = "Courier New" color=red size=6>$$\large  E = y_i - \hat{y_i}$$</font>

<br><br>

- We need to minimize the sum of squares of errors:

$\large \hspace{2cm} SSE = \sum_{i=1}^n(y_i - \hat{y}_i)^2$<br><br>
$\large \hspace{2cm} SSE = \sum_{i=1}^n E_i^2$<br><br>
$\large \hspace{2cm} SSE = E^T E$<br><br>
- Substitute $ E = y - \hat{y}:$<br><br>
$\large \hspace{2cm} SSE = (y - \hat{y})^T (y - \hat{y})$<br><br>
$\large \hspace{2cm} SSE = (y^T - \hat{y}^T) (y - \hat{y}) \hspace{2cm} ........ (i)$<br><br>

<br><br>
- Substitute $ \hat{y} = X\beta$<br><br>
$\large \hspace{2cm} SSE = (y^T - (X \beta)^T) (y - X \beta)$<br><br>
$\large \hspace{2cm} SSE = y^Ty - y^TX \beta - (X \beta)^Ty +  (X \beta)^T)(X \beta)\hspace{2cm} ........ (ii)$<br><br>

<br><br>
- We know that $(A^TB)^T = B^TA$, therefore, in above expression $y^TX \beta = (X \beta)^T)y$

<br>
$\large \hspace{2cm} SSE = y^Ty - y^TX \beta - y^TX \beta  +  \beta^TX^TX\beta$<br><br>
$\large \hspace{2cm} SSE = y^Ty - 2y^TX \beta  +  \beta^TX^TX\beta\hspace{2cm} ........ (iii)$<br><br>

<br><br>
- Now we need to perform matrix differentiation of eq(iii), w.r.t. vector $\beta$ and equate to zero:<br><br>
$\large \hspace{2cm} \frac{\partial}{\partial{\beta}} (y^Ty  - 2y^TX \beta  +  \beta^TX^TX\beta)=0$<br><br>


$\large \hspace{2cm}  \frac{\partial}{\partial{\beta}}(y^Ty) - \frac{\partial}{\partial{\beta}}(2y^TX\beta)+
\frac{\partial}{\partial{\beta}}(\beta^TX^TX\beta) =0 \hspace{2cm} ........ (iv)$<br><br>
<br><br>

**Recap of some formulas of Matrix Differentiation:**
- If **$y$** is $n\times 1$ vector, **$\beta$** is a $m\times 1$ vector and **$X$** is an $n\times m$ matrix. Moreover, if matrix **$X$** is independent of vector **$\beta$**, then we can use the following formulas for matrix differentiation:

$ \hspace{2cm}  y= X\hspace{.2cm}$, then $\large \hspace{2.1cm} \frac{\partial{y}}{\partial{\beta}} \small = 0$<br><br>
$\hspace{2cm} y = X\beta\hspace{.2cm}$, then $\large \hspace{1.8cm} \frac{\partial{y}}{\partial{\beta}} \small = X$<br><br>
$ \hspace{2cm} y = \beta X\hspace{.2cm}$, then $\large \hspace{1.8cm} \frac{\partial{y}}{\partial{\beta}} \small = X^T$<br><br>
$\hspace{2cm} y = \beta^TX\beta\hspace{.2cm}$, then $\large \hspace{1.1cm}\frac{\partial{y}}{\partial{\beta}} \small = 2\beta^TX$<br><br>

$\large \hspace{2cm}  \frac{\partial}{\partial{\beta}}(y^Ty) - \frac{\partial}{\partial{\beta}}(2y^TX\beta)+
\frac{\partial}{\partial{\beta}}(\beta^TX^TX\beta) =0 \hspace{2cm} ........ (iv)$<br><br>
<br><br>
- Differentiating equation (iv), using above formulas, we get:<br><br>
$\large \hspace{2cm}  0 - 2y^TX  + 2 \beta^TX^TX =0 \hspace{2cm} ........ (v)$<br><br>
$\large \hspace{2cm}   2 \beta^TX^TX = 2y^TX$<br><br>
$\large \hspace{2cm}    \beta^TX^TX = y^TX$<br><br>
$\large \hspace{2cm}    \beta^T = y^TX(X^TX)^{-1}$<br><br>

- Taking transpose of both sides:<br>

$\large \hspace{2cm}    (\beta^T)^T = (y^TX(X^TX)^{-1})^T$<br><br>
$\large \hspace{2cm}    \beta = ((X^TX)^{-1})^T (y^TX)^T$<br><br>
$\large \hspace{2cm}    \beta = ((X^TX)^{-1})^T (X^Ty)$<br><br>


- Now $X^TX$ is a symmetric matrix, therefore, its transpose doesn't change  its value:<br>


<font face = "Courier New" color=red size=5>$$\large \hspace{2cm}    \beta = (X^TX)^{-1}X^Ty$$</font><br><br>

<font face = "Courier New" color=magenta  size=5>$$ \large \hat{y} = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \beta_3 X_3 +\cdots + \beta_m X_m$$


>- Taking the derivative of cost function and then solving it for zero to get the set of $\beta$ coefficients becomes computationally expensive when the number of input features are in hundreds or thousands.
>- The Gauss Jordan Elimination technique of computing the inverse of a nxn matrix is $O(n^3)$. So for 999 input features the size of X matrix will be $1000\times1000$, and the complexity of computing the inverse of this matrix will be $10^9$
>- Solution to this problem is `Gradient Descent` in which instead of using a closed form solution, we go slowly towards the answer using approximation technique. (Next Session)

# 5. Example (Simple Linear Regression using Matrices Algebra) 

## a. Load Dataset

```python
import pandas as pd
import numpy as np
df = pd.read_csv("datasets/study-hours.csv")
df
```

## b. Create X matrix and y Vector

```python
sh = np.array(df['study_hours'])
y = np.array(df['gpa'])
sh, y 
```

```python
sh = sh.reshape(sh.size, 1)
sh
```

```python
X = np.hstack((np.ones((sh.size,1)), sh)) # stack arrays horizontally or column-wise
X
```

<font face = "Courier New" color=black size=4>$\hspace{3cm}\large \hspace{2cm}    \beta = (X^TX)^{-1}X^TY$

## c. Calculate the Regression Coefficients

**Using Close Form Solution:**
<font face = "Courier New" color=red size=3>$\hspace{3cm}\large \hspace{2cm}    \beta = (X^TX)^{-1}X^TY$

```python
beta = np.linalg.inv(X.T@X) @ X.T @y
#beta = np.linalg.pinv(X)@y
beta
```

**Using `LinearRegression` Model:**

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(sh, y) #Note that we have just passed the i/p feature as a column vector w/o the addl column of ones
print("Y-intercept (𝛽0): ", model.intercept_)
print("Slope (𝛽1): ", model.coef_)
```

## d. Carry out the Prediction
- What will be the gpa if a student has studied for five hours daily
$$y = \beta_0 + \beta_1 x $$
$$gpa = 0.9 + 0.464*sh $$

```python
studyhours = 5.0
predicted_gpa = beta[0] + beta[1] * studyhours
print("Predicted GPA value for new study hours: {:.4f}".format(predicted_gpa)) # predicted_gpa is a float value
```

**Using `LinearRegression` Model:**

```python
predicted_gpa = model.predict([[studyhours]])
print("Predicted GPA value for new study hours: {:.4f}".format(predicted_gpa[0])) # predicted_gpa is a numpy array
```

## a. Relationship between Output (Sales) and Feature Variables

```python
tv = np.array(df['TV'])
radio = np.array(df['radio'])
newspaper = np.array(df['newspaper'])

x=tv
y = np.array(df['sales'])
numerator = sum(np.multiply(x - np.mean(x), y - np.mean(y)))
denominator = np.multiply((np.sqrt(np.sum((x-np.mean(x))**2))), np.sqrt(np.sum((y-np.mean(y))**2)))
corr_coef_tv = numerator / denominator

x=radio
y = np.array(df['sales'])
numerator = sum(np.multiply(x - np.mean(x), y - np.mean(y)))
denominator = np.multiply((np.sqrt(np.sum((x-np.mean(x))**2))), np.sqrt(np.sum((y-np.mean(y))**2)))
corr_coef_radio = numerator / denominator

x=newspaper
y = np.array(df['sales'])
numerator = sum(np.multiply(x - np.mean(x), y - np.mean(y)))
denominator = np.multiply((np.sqrt(np.sum((x-np.mean(x))**2))), np.sqrt(np.sum((y-np.mean(y))**2)))
corr_coef_newspaper = numerator / denominator


fig, (ax1,ax2, ax3) = plt.subplots(1,3, figsize=(16,6))
ax1.set_title("Pearson Correlation Coefficient: {:.3f}".format(corr_coef_tv))
sns.scatterplot(x='TV', y='sales', data=df, ax=ax1)
ax2.set_title("Pearson Correlation Coefficient: {:.3f}".format(corr_coef_radio))
sns.scatterplot(x='radio', y='sales', data=df, ax=ax2)
ax3.set_title("Pearson Correlation Coefficient: {:.3f}".format(corr_coef_newspaper))
sns.scatterplot(x='newspaper', y='sales', data=df, ax=ax3);
```

### (i) Mean Absolute Error (MAE):
- Absolute means we do not consider the sign, i.e., $ d_i = |y_i - \hat{y_i}|$
$$ \large MAE =  \frac{1}{n}\sum_{i=1}^n|y_i - \hat{y}| \hspace{5 cm}$$
- The **advantages** of MAE are that the unit is same and it works well in case of outliers as well.
- The **disadvantage** is that the modulus function is not differentiable.

### (ii) Mean Squared Error (MSE):
- To overcome the limitation of MAE, i.e., to make the cost function differentiable
- In Mean Squared Error (MSE), instead of taking the modulus, we take the mean of all the residual errors and take their average:
$$ \large MSE  =  \frac{1}{n}\sum_{i=1}^n(y_i - \hat{y})^2 \hspace{5 cm}$$

- The **advantages** of MSE is that the cost function is differentiable.
- The **limitation** is that the unit are different. For example, if the unit of output variable ($y$) is Rs, then the MSE unit is RS$^2$. So if MSE of a point is 10 that means the actual error is square of 10. 
- The other limitation is that it doesn't work well in case of outliers, because in case of a outlier the error is squared and is a big value. So the loss function is effected much by outliers. So if there are lot of outliers in yor dataset, you should go with MAE instead of MSE.

### (iii) Root Mean Squared Error (RMSE):
$$ \large RMSE = \sqrt{MSE}  = \sqrt{\frac{1}{n}\sum_{i=1}^n(y_i - \hat{y})^2} \hspace{5 cm}$$
- The **advantages** is that like MSE, this cost function is also differentiable. Main advantage is that the unit is same, which is not the case of MSE.
- The **disadvantage** is, that it is not that robust to outliers. (We use RMSE in case of deep learning)


**Using above Formulae:**

```python
ybar = np.mean(y)
yhat = betas[0] + betas[1]*df['TV'] + betas[2]*df['radio'] +  betas[3] * df['newspaper']
mae = (np.sum(abs(y - yhat)))/len(y)
mse = (np.sum((y - yhat)**2))/len(y)
rmse = np.sqrt((np.sum((y - yhat)**2))/len(y))

print("MAE: {:.3f}".format(mae))
print("MSE: {:.3f}".format(mse))
print("RMSE: {:.3f}".format(rmse))
```

### (iv) $R^2$ Score:

<img align="center" width="1000"  src="images/r2score.png"  >


$$ \large R^2 =  1 - \frac{SSE}{SST}\hspace{14cm}$$

- Where,<br>

$\hspace{3cm} SSE =  \sum_{i=1}^n(y_i - \hat{y}_i)^2$<br>
$ \hspace{3cm}SSR =  \sum_{i=1}^n(\hat{y}_i - \bar{y})^2$<br>
$ \hspace{3cm}SST =  \sum_{i=1}^n(y_i - \bar{y})^2$<br>


**Using above Formulae:**

```python
ybar = np.mean(y)
sse = np.sum((y - yhat)**2)
sst = np.sum((y - ybar)**2)
r2 = 1 - (sse/sst)
print("R2 Score: {:.4f}".format(r2))
```

### (v) Adjusted $R^2$ Score:**
- The limitation or R2 score is that, if you add more features to your model, the R2 score will either increase or stay the same, but will never decrease.
- Adjusted R2 score is a modified form of R2 score, whose value increases if new predictors tend to improve model's performance and decreases if new predictors do not improve performance as expected.

<center><font face = "Comic sans MS" color=green size=4>$ \large R_a^2 =  \small 1 - \large [\frac{(1-R^2)(n-1)}{n-m-1}]$</font></center><br><br>

<center>$ \large R_a^2 =  \small 1 - (\large \frac{n-1}{SST})\small \small MSE\hspace{1cm}$</font></center><br><br>
<center>$ \large R_a^2 =  \small R^2 - \large \frac{m(1-R^2)}{n-m-1}\hspace{1.5cm}$</font></center><br><br>

>- where n is number of observations in sample and m is number of features/predictors/independent variables in model
>- All the three formulas give you the same result, actually if your reorganize them algebbraically you can end up with the same formula

```python
n = len(X)
m = 3
r2_adj = 1 - ((1-r2)*(n-1)/(n-m-1))
print("R2 Score: {:.4f}".format(r2_adj))
```

<font face = "Garamond" color=green size=5>**c. Homoscedasticity:**
- It is a condition where the variance of the residual errors in a regression model is constant.
- This can be checked using residual plot, that shows the residual errors on the vertical axis and the independent variable on the horizontal axis. If the residuals do not fan out in a triangular fashion that means that the homoscedasticity (equal variance) assumption is met
- Violation of this assumption makes your coefficients less accurate
<img align="center" width="1500"  src="images/homoscedasticity.png"  >
    
    
- `What to do if this assumption is violated?`   
    - Transform the dependent variable. One common transformation is to simply take the log of the dependent variable. For example, if we are using population size (independent variable) to predict the number of flower shops in a city (dependent variable), we may instead try to use population size to predict the log of the number of flower shops in a city. Using the log of the dependent variable, rather than the original dependent variable, often causes heteroskedasticity to go away.
    - Redefine the dependent variable.  One common way to redefine the dependent variable is to use a rate, rather than the raw value. For example, instead of using the population size to predict the number of flower shops in a city, we may instead use population size to predict the number of flower shops per capita. In most cases, this reduces the variability that naturally occurs among larger populations since we’re measuring the number of flower shops per person, rather than the sheer amount of flower shops.
    - Use weighted regression. Another way to fix heteroscedasticity is to use weighted regression. This type of regression assigns a weight to each data point based on the variance of its fitted value. Essentially, this gives small weights to data points that have higher variances, which shrinks their squared residuals. When the proper weights are used, this can eliminate the problem of heteroscedasticity.
